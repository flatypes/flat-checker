```python
def count_lines(file_path: str) -> tuple[int, int, int]:
  with open(file_path, 'r') as f:
    lines = f.readlines()

  loc = 0
  lot = 0  # type annotations
  loi = 0  # invariants
  type_start = False
  docstring_start = False
  for line in lines:
    line = line.strip()
    if line.startswith("'''") or line.startswith('"""'):
      docstring_start = True

    if docstring_start:
      if line.endswith("'''") or line.endswith('"""'):
        docstring_start = False
      continue

    if not line or line.startswith('#'):  # skip empty lines and comments
      continue

    if type_start and line.startswith("r'"):
      lot += 1
      if line.endswith(')'):
        type_start = False
      continue

    type_start = False
    if line.startswith('type') and 'lang' in line:
      type_start = True
      lot += 1
    elif line.startswith('inv'):
      loi += 1
    else:
      loc += 1

  return loc, lot, loi
```

```python
import os
import statistics as stat

panini_loc = []
panini_lot = []
panini_loi = []

for f in os.listdir('../examples/panini-bench'):
  if f.endswith('.py'):
    loc, lot, loi = count_lines(os.path.join('../examples/panini-bench', f))
    panini_loc.append(loc)
    panini_lot.append(lot)
    panini_loi.append(loi)

print('Panini Benchmarks Total:')
print(f'LoC: {sum(panini_loc)}')
print(f'LoT: {sum(panini_lot)}')
print(f'LoI: {sum(panini_loi)}')
print(f'LoA/LoC: {(sum(panini_lot) + sum(panini_loi)) / sum(panini_loc):.2f}')

print('\nLines of code:')
print(f'min: {min(panini_loc)}')
print(f'max: {max(panini_loc)}')
print(f'median: {stat.median(panini_loc)}')
print(f'11+: {sum(1 for x in panini_loc if x > 10)}')

print(f'\n2 type annotations: {sum(1 for x in panini_lot if x == 2)}')
print(f'1 invariants: {sum(1 for x in panini_loi if x == 1)}')
print(f'2 invariants: {sum(1 for x in panini_loi if x == 2)}')
print(f'3 invariants: {sum(1 for x in panini_loi if x == 3)}')
print(f'4+ invariants: {sum(1 for x in panini_loi if x >= 4)}')

print('\nPer File:')
print(f'LoC: {stat.mean(panini_loc):.2f}')
print(f'LoT: {stat.mean(panini_lot):.2f}')
print(f'LoI: {stat.mean(panini_loi):.2f}')
```
